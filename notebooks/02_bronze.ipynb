```python
from pyspark.sql import functions as F

RAW_PATH = "/Volumes/workspace/metropulse/raw/gtfs"
SCHEMA = "workspace.metropulse"

files = [f.name for f in dbutils.fs.ls(RAW_PATH) if f.name.endswith(".txt")]

for filename in files:
    table_name = f"{SCHEMA}.bronze_{filename.replace('.txt', '')}"

    df = (
        spark.read
        .option("header", True)
        .option("inferSchema", False)
        .csv(f"{RAW_PATH}/{filename}")
        .withColumn("_source_file", F.lit(filename))
        .withColumn("_ingested_at", F.current_timestamp())
    )

    df.write.mode("overwrite").option("overwriteSchema", True).saveAsTable(table_name)

    print(f"{table_name:50} {spark.table(table_name).count():>10,} rows")
```
